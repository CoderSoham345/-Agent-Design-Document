## Honest Validation

For this audit, I re-evaluated my Week 5 model using a more careful validation strategy.

The original model used a random train/test split. While this provides a useful baseline, it can sometimes produce optimistic estimates if similar records appear in both training and testing.

An honest validation strategy, such as a grouped split or time-aware split when appropriate, provides a more realistic estimate of how the model may perform on unseen data.

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd

df = pd.read_csv("refresh_queue_sample.csv")

print(df.head())
print(df.columns)
```

```python
from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(test_size=0.2, random_state=42)

train_idx, test_idx = next(
    gss.split(df, groups=df["client_id"])
)

print("Training rows:", len(train_idx))
print("Testing rows:", len(test_idx))
print("Unique clients:", df["client_id"].nunique())
```

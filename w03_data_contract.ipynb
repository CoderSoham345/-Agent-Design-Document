## Unit of Analysis

One row represents a single page for a single day from the Google Search Console warehouse dataset. Every row contains the historical search performance metrics for one page on one specific date.

## Tables Used

I use the Search Console warehouse table containing the daily page-level search metrics. The analysis focuses on historical search performance data available before the prediction date.

## Time Window

The analysis uses **March 2026 (2026-03)** as the working month.

I intentionally avoid using **June 2026**, because it is the final month in the released warehouse and should remain an unseen evaluation period instead of being used during feature development.

The objective is to predict or rank future page search performance using only information that is available before the prediction time.

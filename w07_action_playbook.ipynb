## Ranked Actions + Reason Codes

### High Priority
- Refresh content with high refresh scores.
- Update outdated information and statistics.
- Improve titles, meta descriptions, and internal links.

Reason Codes
- High Refresh Score
- Declining Traffic
- Low CTR
- Low Engagement

### Medium Priority
- Expand thin content.
- Add FAQs and supporting sections.
- Improve keyword coverage.

Reason Codes
- Moderate Refresh Score
- Moderate Traffic Decline

### Low Priority
- Monitor content performance.
- Review after future data updates.

Reason Codes
- Stable Performance
- Low Business Impact

## Intended Use

This playbook is designed as a decision-support tool for SEO and content teams. It ranks existing content by estimated refresh priority and provides reason codes explaining why a page may benefit from review.

The model is intended to help prioritize manual work rather than automate publishing decisions.

## Limits

- Predictions are based on historical content performance.
- Results should be interpreted as directional, not guaranteed outcomes.
- Changes in search algorithms, seasonality, or business priorities may reduce accuracy.
- The model has been validated on available data only and should not be treated as a production system.

- Current traffic trends
- Search intent
- Content quality

## What Should NOT Be Automated

- Deleting pages automatically
